```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/student_placement_prediction_dataset_2026.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python

# TASK 1: Understand the dataset

print("FIRST 5 STUDENT RECORDS")
display(df.head())

print("LAST 5 STUDENT RECORDS")
display(df.tail())

print("5 RANDOM STUDENT RECORDS")
display(df.sample(5, random_state=42))

print("\nDATASET DIMENSIONS")
print("Total students:", df.shape[0])
print("Total columns:", df.shape[1])

print("\nCOLUMN NAMES")
print(df.columns.tolist())

# TASK 2: Basic data inspection

print("\nDATA TYPES AND NON-NULL COUNTS")
df.info()

print("\nMISSING VALUES IN EACH COLUMN")
print(df.isnull().sum())

print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

print("\nNUMERICAL SUMMARY STATISTICS")
display(df.describe())

print("\nUNIQUE VALUES IN EACH COLUMN")
print(df.nunique())
```

```python

# TASK 2: Analyze placement status

print("Unique placement status values:")
print(df["placement_status"].unique())

print("\nCount of each placement status:")
print(df["placement_status"].value_counts(dropna=False))

print("\nPlacement status percentages:")
print(
    (df["placement_status"].value_counts(normalize=True, dropna=False) * 100)
    .round(2)
)

# Visualize the placement status distribution
df["placement_status"].value_counts(dropna=False).plot(
    kind="bar",
    figsize=(7, 4),
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Student Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# TASK 3: Data Cleaning and Preprocessing

# Make a copy of the original dataset
df_clean = df.copy()

# Remove duplicate rows
duplicates_before = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()

print("Duplicate rows removed:", duplicates_before)

# Remove student_id because it is only an identifier
if "student_id" in df_clean.columns:
    df_clean = df_clean.drop(columns=["student_id"])

# Remove salary because it is known after placement
if "salary_package_lpa" in df_clean.columns:
    df_clean = df_clean.drop(columns=["salary_package_lpa"])

print("\nRemaining missing values:")
print(df_clean.isnull().sum())

print("\nCleaned dataset shape:", df_clean.shape)

print("\nRemaining columns:")
print(df_clean.columns.tolist())

display(df_clean.head())
```

```python

# TASK 3: Check data quality

print("Missing values before handling:")
print(df_clean.isnull().sum())

print("\nTotal missing values:")
print(df_clean.isnull().sum().sum())

print("\nDuplicate rows remaining:")
print(df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)

# Handle missing values if any are present
for column in df_clean.columns:
    if df_clean[column].isnull().any():
        if pd.api.types.is_numeric_dtype(df_clean[column]):
            df_clean[column] = df_clean[column].fillna(
                df_clean[column].median()
            )
        else:
            mode_values = df_clean[column].mode()
            if not mode_values.empty:
                df_clean[column] = df_clean[column].fillna(
                    mode_values.iloc[0]
                )
            else:
                df_clean[column] = df_clean[column].fillna("Unknown")

print("\nMissing values after handling:")
print(df_clean.isnull().sum().sum())

print("\nFinal cleaned dataset shape:", df_clean.shape)
```

```python

# TASK 6: Categorical Feature Analysis

categorical_features = [
    "gender",
    "branch",
    "college_tier",
    "volunteer_experience"
]

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

for feature, ax in zip(categorical_features, axes.flatten()):
    placement_rates = pd.crosstab(
        df_clean[feature],
        df_clean["placement_status"],
        normalize="index"
    ) * 100

    placement_rates.plot(
        kind="bar",
        stacked=True,
        ax=ax
    )

    ax.set_title(f"{feature.replace('_', ' ').title()} vs Placement")
    ax.set_ylabel("Percentage of Students")
    ax.set_xlabel(feature.replace("_", " ").title())
    ax.legend(title="Placement Status")
    ax.tick_params(axis="x", rotation=30)

plt.suptitle("Categorical Features and Placement Rates", fontsize=16)
plt.tight_layout()
plt.show()

print("Categorical feature analysis completed.")
```

```python

# TASK 7: Separate features and target

X = df_clean.drop(columns=["placement_status"])
y = df_clean["placement_status"].astype(str).str.strip().str.lower()

# Convert target labels into numerical values
target_mapping = {
    "placed": 1,
    "not placed": 0
}

y = y.map(target_mapping)

if y.isnull().any():
    raise ValueError(
        "Unexpected placement_status values found. "
        "Check df_clean['placement_status'].unique()."
    )

y = y.astype(int)

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts().rename(index={1: "Placed", 0: "Not Placed"}))

print("\nInput features:")
print(X.columns.tolist())

print("\nTarget mapping: Placed = 1, Not Placed = 0")
```

```python

# TASK 8: Train-test split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

```python

# TASK 9: Feature Encoding and Scaling

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Identify numerical and categorical columns
numeric_features = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Prepare numerical features
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Prepare categorical features
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing methods
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# Learn preprocessing rules from training data only
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)
print("Original testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)

print("\nNumerical features:", numeric_features)
print("Categorical features:", categorical_features)
print("\nEncoding and scaling completed successfully!")
```

```python

# TASK 10: Train four machine learning models

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.base import clone

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000, random_state=42
    ),
    "Random Forest": RandomForestClassifier(
        n_estimators=100, random_state=42, n_jobs=-1
    ),
    "Support Vector Machine": LinearSVC(
        random_state=42, max_iter=5000
    ),
    "K-Nearest Neighbors": KNeighborsClassifier(
        n_neighbors=5, n_jobs=-1
    )
}

trained_models = {}

for name, model in models.items():
    print(f"Training {name}...")

    pipeline = Pipeline(steps=[
        ("preprocessor", clone(preprocessor)),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)
    trained_models[name] = pipeline

print("\nAll four models have finished training!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
import pandas as pd

results = []

for name, pipeline in trained_models.items():
    print(f"\nEvaluating: {name}")

    y_pred = pipeline.predict(X_test)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0)
    })

    print(classification_report(
        y_test,
        y_pred,
        target_names=["Not Placed", "Placed"],
        zero_division=0
    ))

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(
    by="F1 Score", ascending=False
).reset_index(drop=True)

print("\nMODEL COMPARISON")
display(results_df.round(4))
```

```python

from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(12, 9))
axes = axes.flatten()

for ax, (name, pipeline) in zip(axes, trained_models.items()):
    y_pred = pipeline.predict(X_test)
    cm = confusion_matrix(y_test, y_pred, labels=[0, 1])

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=["Not Placed", "Placed"],
        yticklabels=["Not Placed", "Placed"],
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted Status")
    ax.set_ylabel("Actual Status")

plt.tight_layout()
plt.show()
```

```python

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Use a smaller training sample to make tuning manageable
sample_size = min(20000, len(X_train))
X_tune = X_train.sample(n=sample_size, random_state=42)
y_tune = y_train.loc[X_tune.index]

# Build a pipeline using the same preprocessing
tuning_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(random_state=42, n_jobs=-1))
])

# Try a small set of parameter combinations
param_grid = {
    "model__n_estimators": [100, 150],
    "model__max_depth": [None, 15],
    "model__min_samples_split": [2, 5]
}

grid_search = GridSearchCV(
    estimator=tuning_pipeline,
    param_grid=param_grid,
    scoring="f1",
    cv=3,
    n_jobs=-1,
    verbose=1
)

print("Starting hyperparameter tuning...")
grid_search.fit(X_tune, y_tune)

print("\nBest parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation F1 score:")
print(round(grid_search.best_score_, 4))

# Evaluate the tuned model on the untouched test set
best_model = grid_search.best_estimator_
tuned_predictions = best_model.predict(X_test)

print("\nTUNED MODEL TEST RESULTS")
print("Accuracy:", round(accuracy_score(y_test, tuned_predictions), 4))
print("Precision:", round(precision_score(y_test, tuned_predictions, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, tuned_predictions, zero_division=0), 4))
print("F1 Score:", round(f1_score(y_test, tuned_predictions, zero_division=0), 4))
```

```python

# Select the tuned model for the prediction system
final_model = best_model

print("Final model selected successfully!")
print("Model type:", type(final_model.named_steps["model"]).__name__)

# Check whether probability predictions are supported
if hasattr(final_model.named_steps["model"], "predict_proba"):
    print("Probability prediction: Supported")
else:
    print("Probability prediction: Not supported")
```

```python

import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd

# Create input fields based on the dataset
input_widgets = {}

for column in X.columns:
    if pd.api.types.is_numeric_dtype(X[column]):
        input_widgets[column] = widgets.FloatText(
            value=float(X[column].median()),
            description=column,
            style={"description_width": "initial"},
            layout=widgets.Layout(width="500px")
        )
    else:
        options = sorted(
            X[column].dropna().astype(str).unique().tolist()
        )
        input_widgets[column] = widgets.Dropdown(
            options=options,
            value=options[0] if options else None,
            description=column,
            style={"description_width": "initial"},
            layout=widgets.Layout(width="500px")
        )

predict_button = widgets.Button(
    description="Predict Placement",
    button_style="success",
    icon="check"
)

output = widgets.Output()

def predict_placement(button):
    with output:
        clear_output()

        # Collect values entered by the user
        student_data = {
            column: [widget.value]
            for column, widget in input_widgets.items()
        }
        student_df = pd.DataFrame(student_data, columns=X.columns)

        # Predict placement
        prediction = int(final_model.predict(student_df)[0])

        # Estimate probability, if supported
        model = final_model.named_steps["model"]

        print("STUDENT PLACEMENT PREDICTION")
        print("-" * 35)

        if hasattr(final_model, "predict_proba"):
            probabilities = final_model.predict_proba(student_df)[0]
            class_index = list(final_model.named_steps["model"].classes_).index(1)
            probability = float(probabilities[class_index]) * 100

            print(f"Estimated placement probability: {probability:.2f}%")

            if probability >= 70:
                print("Model category: Higher estimated probability")
            elif probability >= 40:
                print("Model category: Moderate estimated probability")
            else:
                print("Model category: Lower estimated probability")
        else:
            print("Probability estimate is not supported by this model.")

        print(
            "Predicted status:",
            "Placed" if prediction == 1 else "Not Placed"
        )
        print("\nThis is a model estimate, not a guarantee of actual placement.")

predict_button.on_click(predict_placement)

form = widgets.VBox(list(input_widgets.values()))
display(form, predict_button, output)
```
